```python
from sklearn.linear_model import LogisticRegression

X = [[2], [3], [5], [7], [8]]
y = [0, 0, 1, 1, 1]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[6]])
probability = model.predict_proba([[6]])

print("Prediction:", prediction)
print("Probability:", probability)
```
